Bucles For

```python
# Los bucles for generalmente se usan para recorrer una secuencia de objetos. El bloque de codigo que contiene el bucle se ejecutara por cada objeto que exista en la secuencia ----> ITERACION
estudiantes_notas = [("Camila",5.0),("Lorena",4.0),("Lina",4.2),("Adrian",5.0),("Edwin",3.8),("Andrea",4.7)]

for i in estudiantes_notas: # Para cada tupla estudiante - nota en la lista de estudiantes_notas, realice las siguientes acciones
    print(f"{i[0]} obtuvo una nota de {i[1]} en el diplomado de análisis de datos")
    print("---------------------------------------------------------------------")
```

```python
suma_total = 0 #Declaro la variable suma_total que almacenara el acumulado de la suma

# En el rango es: inicia en 10 termina en 51 y va de 5 en 5 
for numero in range(10,51,5):  # Ejemplo iterando sobre un rango de numeros. Se suma cada número y se imprime el resultado total.
  print(f"Imprimiendo el número:  {numero}")
  suma_total += numero #Esto es lo mismo que numero = numero + 5 por el 5 en 5

print("------------------------------------------")
print(f"La sumatoria total es: {suma_total}")
```

Bucle While

```python
#El bucle while evalua una condición y ejecuta el bloque de código interno solamente si la condicion es verdadera. Se detendrá cuando la condición cambie a Falso
contador = 0
while contador <= 5:
  print(f"El valor del contador es: {contador}")
  contador += 1 # Indice = indice + 1
```

Break / Continue
break:
Detiene completamente el ciclo cuando se cumple una condición.
El programa sale del ciclo y continúa con el código que viene después.
continue:
Omite la iteración actual cuando se cumple una condición.
El ciclo NO termina; simplemente pasa a la siguiente iteración.

```python
# Break

for numero in range(1, 11):

    if numero == 5:
        break

    print(numero)
```

```python
estudiantes = ['Yis', 'David', 'Jennifer','Juana']
for estudiante in estudiantes:
  if estudiante == 'David':
    continue # Omite pero no detiene el ciclo
  print(f"Hola {estudiante}")
```

```python
suma_total = 0 #Declaro la variable suma_total que almacenara el acumulado de la suma

# Uso de las palabras claves break y continue

for numero in range(102):  # Ejemplo iterando sobre un rango de numeros del 0 al 10.
  if (numero % 2 == 0): # Si el número es par, ejecuta el siguiente bloque de código
    print(f"El número {numero} es par")
  else: # Si el número es impar, no debe imprimir nada
    continue  #Pasa al siguiente número

  suma_total += numero

  if(numero == 12):
    break # Sale del bucle cuando llegué al número 8

print("------------------------------------------")
print(f"La sumatoria total es: {suma_total}")
```

Parte 1: Bucle for con edades
Crea una lista de edades: [12, 17, 8, 15, 22, 9, 30, 25].

Usa un bucle for para iterar sobre cada edad:

Si la edad es 25, imprime "¡Encontramos al estudiante de 25 años! Deteniendo el análisis." y termina el bucle con break.

Si la edad es menor a 10, omite esa iteración con continue.

Si la edad es 18 o mayor, imprime "Adulto: [edad] años".

En cualquier otro caso, imprime "Menor: [edad] años".

```python
lista_edades = [12, 17, 8, 15, 22, 9, 30, 25]

for edad in lista_edades:
  if edad == 25:
    print(f"¡Encontramos al estudiante de 25 años!")
    break
  elif edad < 10:
    continue # Omite pero no detiene el ciclo
  elif edad >=18:
    print(f"Adulto: {edad} años")

  else:
    print(f"Menor: {edad} años")
```

Parte 2: Bucle while con nombres
Crea una lista de nombres: ["Juan", "Pedro", "Jorge", "María","Ana"].

Usa un bucle while para iterar sobre los nombres:

Si el nombre es "Ana", imprime "¡Ana está en la lista!" y termina el bucle con break.

Si el nombre empieza con "J", omite ese nombre con continue.

En cualquier otro caso, imprime el nombre.

```python
lista_nombres = ["Juan","Pedro","Jorge","María","Ana"]

i = 0
while i<len(lista_nombres): ##len funcion que permite contar
    nombre = lista_nombres[i]
    i +=1
     
    if nombre == "Ana":
        print(f"¡Ana está en la lista!")
        break
    elif nombre[0] == "J":
        continue
    else:
        print(nombre) 
```
