Construye una figura con **cuatro gráficos en una cuadrícula de 2 × 2**, con una apariencia coherente y lista para presentar.

Requisitos:
- Usa `plt.subplots(2, 2, figsize=(...))`.
- Incluye al menos un histograma, un diagrama de caja y un gráfico de dispersión.
- Usa títulos que expliquen qué se está comparando.
- Aplica una paleta consistente y reduce el ruido visual.


Plantilla para completar:

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

try:
    iris
except NameError:
    iris = sns.load_dataset("iris")

sns.set_theme(
    style="whitegrid",
    context="notebook",
    palette="deep",
    font_scale=1.0
)
plt.rcParams.update({
    "figure.dpi": 110,
    "axes.titleweight": "bold",
    "axes.titlepad": 12,
    "axes.labelcolor": "#344054",
    "axes.titlecolor": "#1D2939",
    "grid.alpha": 0.22,
    "axes.spines.top": False,
    "axes.spines.right": False
})

display(iris.head())
```

```python
colores = ["#2A788E", "#6A8E7F", "#E07A5F", "#8064A2"]

fig, axes = plt.subplots(2, 2, figsize=(12, 10))
fig.suptitle("Análisis Iris",
             fontsize=16, fontweight="bold", x=0.05, ha="left")

#Grafico 1: Histograma
sns.histplot(data=iris, x="sepal_length", hue="species", color=colores[0],
             alpha=0.75, line_kws={"linewidth": 2.5}, multiple="stack", ax=axes[0,0])
axes[0,0].set_title("Longitud del sépalo por especie")
axes[0,0].set_xlabel("Longitud (cm)")
axes[0,0].set_ylabel("Frecuencia")

#Grafico 2: Diagrama de caja
sns.boxplot(data=iris, x="species", y="sepal_width", color=colores[1], ax=axes[0,1])
axes[0,1].set_title("Ancho del sepalo por especie")
axes[0,1].set_xlabel("Especie")
axes[0,1].set_ylabel("Longitud (cm)")

#Grafico 3: Dispersion
sns.scatterplot(data=iris, x="petal_length", y="petal_width", hue="species",
                color=colores[1], ax=axes[1,0], s=70 )
axes[1,0].set_title("Relacion entre largo y ancho del petalo")
axes[1,0].set_xlabel("Largo petalo (cm)")
axes[1,0].set_ylabel("Ancho petalo (cm)")

#Grafico 4: barras
sns.barplot(data=iris, x="species", y="sepal_width", estimator='mean',
            color=colores[2], ax=axes[1,1])
axes[1,1].set_title("Ancho del sépalo por especie")
axes[1,1].set_xlabel("Especie")
axes[1,1].set_ylabel("Longitud (cm)")


sns.despine()
plt.tight_layout()
plt.show()
```
