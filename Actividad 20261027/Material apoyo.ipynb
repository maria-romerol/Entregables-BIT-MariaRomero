## Introducción a Matplotlib


Guía conceptual: qué gráfica usar y cuándo

| Necesidad analítica | Pregunta de negocio | Gráfica recomendada |
|---|---|---|
| Contar categorías | ¿Cuántos pedidos hay por restaurante? |  barras |
| Comparar métricas | ¿Qué restaurante tarda más? | Barras con promedio |
| Ver distribución | ¿Cómo se comportan los tiempos de entrega? | Histograma |
| Detectar outliers | ¿Hay pedidos extremadamente lentos? | diagrama de caja y bigotes |
| Ver evolución | ¿Cómo cambian los pedidos por hora? | Línea |
| Relacionar variables | ¿A mayor distancia, mayor duración? | diagrama de dispersión |
| Ver correlaciones | ¿Qué variables se mueven juntas? | Heatmap de correlación |
| Ver composición | ¿Qué porcentaje usa cada método de pago? | Barras apiladas / pie con pocas categorías |
| Priorizar | ¿Qué ubicaciones concentran pedidos? | Pareto |
| Explorar interactivamente | ¿Qué punto específico tiene más retraso? | Plotly (librería de python)|

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
import pandas as pd
df_line = pd.DataFrame({
    'Mes': ["Enero","Febrero","Marzo","Abril","Mayo","Junio","Julio","Agosto","Septiembre","Octubre","Noviembre", "Diciembre"],
    'Ventas': [10, 35, 24, 50, 74, 80, 64, 95, 110, 104, 88, 120]
})
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

plt.figure(figsize=(6, 4)) ## Tamaño de las figura
plt.plot(df_line['Mes'], df_line['Ventas']) ## eje x, eje y
plt.title("Ventas 2025")
plt.xlabel("Mes")
plt.ylabel("Promedio Ventas (Millones $COP)")
plt.xticks(df_line['Mes'], rotation=45) # Ajuste en los nombres del eje x, rotarlos para que se puedan leer
plt.grid()
plt.show()
```

```python
# Gráfico de líneas con Seaborn
plt.figure(figsize=(10, 6))
sns.lineplot(data=df_line, x='Mes', y='Ventas', marker='s')
# Añadir etiquetas a la grafica
for i, txt in enumerate(df_line['Ventas']):
    plt.text(df_line['Mes'][i], df_line['Ventas'][i], txt, ha='right', va='bottom')
plt.yticks(df_line['Ventas'], rotation=35)
plt.xticks(df_line['Mes'], rotation=45)
plt.title("Ventas 2024 con Seaborn")
plt.grid()
plt.show()

# ha = horizontal alignment: Alinea el texto horizontalmente respecto al punto (izquierda, centro o derecha) 'left', 'center', 'right'
# va = 	vertical alignment: Alinea el texto verticalmente respecto al punto (arriba, centro o abajo) 'top', 'center', 'bottom', 'baseline'
# baseline, texto se alinea por su linea base
# bottom, la parte mas baja del texto queda en el punto
# center, centro vertical del texto queda en el punto
# top, la parte mas alta del texto queda en el punto
```

## Cargar datos IRIS

```python
iris = sns.load_dataset("iris")
```

```python
iris
```

### 2. Gráfico de Dispersión (Scatterplot)
Muestra la relación entre dos variables numéricas continuas.

```python
plt.figure(figsize=(6, 4))
for species in iris["species"].unique():
    subset = iris[iris["species"] == species]
    plt.scatter(subset["sepal_length"], subset["sepal_width"], label=species)
plt.legend()
plt.title("Gráfico de Dispersión con Matplotlib")
plt.xlabel("Sepal Length (mm)")
plt.ylabel("Sepal Width (mm)")
plt.show()
```

```python
#Seaborn
plt.figure(figsize=(6, 4))
sns.scatterplot(data=iris, x="sepal_length", y="sepal_width", hue="species")
plt.title("Gráfico de Dispersión con Seaborn")
plt.show()
```

### 3. Gráfico de Barras
Muestra comparaciones entre categorías.


```python
df_grouped = iris.groupby("species")["petal_length"].median()
plt.figure(figsize=(6, 4))
plt.bar(df_grouped.index, df_grouped.values)
plt.title("Gráfico de Barras con Matplotlib")
plt.xlabel("Especies")
plt.ylabel("Largo del Pétalo (mm)")
plt.show()
```

```python
plt.figure(figsize=(6, 4))
sns.barplot(data=iris, x="species", y="sepal_width", estimator='mean', palette = "pastel")
plt.title("Gráfico de Barras con Seaborn")
plt.show()
```

### 4. Gráfico de Torta (Pie Chart)
Muestra proporciones de una variable categórica.

```python
df_species = iris['species'].value_counts()
plt.figure(figsize=(6, 6))
plt.pie(df_species, labels=df_species.index, autopct='%1.3f%%', startangle=0) # valores numericos, valores por categoria, n° decimales, angulo del primer segmento
plt.title("Distribución de Especies (Matplotlib)")
plt.show()
```

```python
# Seaborn no tiene gráficos de torta nativos, pero se pueden usar gráficos de barras proporcionales
plt.figure(figsize=(6, 4))
sns.barplot(x=df_species.index, y=df_species.values / df_species.values.sum())
plt.title("Distribución de Especies (Seaborn - Aproximación)")
plt.ylabel("Proporción")
plt.show()
```

### 5. Histograma
Muestra la distribución de una variable numérica.

```python
plt.figure(figsize=(6, 4))
plt.hist(iris['sepal_length'].dropna(), bins=20, edgecolor='black', alpha=0.9) # datos, eliminación valores nulos, división de los datos, color del borde, transparencia del grafico
plt.title("Histograma con Matplotlib")
plt.xlabel("Sepal Length (mm)")
plt.ylabel("Frecuencia")
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(6, 4))
sns.histplot(iris['sepal_length'].dropna(), bins=20)
plt.title("Histograma de Sepal Length con Seaborn")
plt.xlabel("Sepal Length (mm)")
plt.ylabel("Frecuencia")
plt.grid(True)
plt.show()
```

### 6. Gráfico de Caja (Boxplot)
Muestra la distribución y valores atípicos de una variable numérica.

```python
plt.figure(figsize=(6, 4))
iris.boxplot(column="petal_length", by="species")
plt.title("Boxplot con Matplotlib")
plt.suptitle("")
plt.xlabel("Especies")
plt.ylabel("Largo del Pétalo (mm)")
plt.show()
```

```python
plt.figure(figsize=(6, 4))
sns.boxplot(data=iris, x="species", y="petal_length")
plt.title("Boxplot de Largo del Pétalo por Especie con Seaborn")
plt.grid()
plt.show()
```
