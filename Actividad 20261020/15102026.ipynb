Un `array` es una estructura que permite almacenar y procesar datos. Puede tener una dimensión (como una secuencia) o varias dimensiones (como una tabla numérica).

```python
ventas = np.array([120, 150, 90, 200, 180]) 
#Se toma como una matriz matematica

print("Ventas:", ventas)
print("Tipo de objeto:", type(ventas))
print("Forma (shape):", ventas.shape)
print("Número de dimensiones (ndim):", ventas.ndim) #dimensiones de la matri
print("Tipo de dato (dtype):", ventas.dtype)
print("Cantidad de elementos (size):", ventas.size)
```

- `shape` indica el tamaño de cada dimensión.
- `ndim` indica cuántas dimensiones tiene el array.
- `dtype` muestra el tipo de dato almacenado.
- `size` indica el número total de elementos.

La `vectorización` permite aplicar una operación a todos los elementos de un array sin escribir un ciclo explícito.

Las `estadísticas descriptivas` ayudan a resumir el comportamiento de los datos. En este ejercicio analizaremos las ventas de una semana.

```python
ventas_semana = np.array([120, 150, 90, 200, 180, 160, 140])

print("Total vendido:", np.sum(ventas_semana)) #SUMATORIA
print("Promedio:", np.mean(ventas_semana)) #PROMEDIO
print("Mediana:", np.median(ventas_semana)) #MEDIANA
print("Desviación estándar:", np.std(ventas_semana)) #DESVIACION ESTANDAR
print("Mínimo:", np.min(ventas_semana)) #MINIMO
print("Máximo:", np.max(ventas_semana)) #MAX
print("Percentil 75:", np.percentile(ventas_semana, 75)) #PERCENTILES
```

- **Media:** promedio de los valores.
- **Mediana:** valor central cuando los datos se ordenan.
- **Desviación estándar:** mide la dispersión de los valores alrededor de la media.
- **Percentil 75:** valor que deja aproximadamente el 75 % de las observaciones por debajo o igual a él.

Una máscara booleana es un array de valores `True` y `False` que permite seleccionar elementos que cumplen una condición.


```python
ventas = np.array([120, 150, 90, 200, 180, 160, 140, 250, 80, 190])

mascara = ventas > 150

print("Máscara:", mascara)
print("Ventas superiores a 150:", ventas[mascara])
print("Cantidad de días con ventas superiores a 150:", np.sum(ventas > 150))
print("Porcentaje de días con ventas superiores a 150:", np.mean(ventas > 150) * 100)
```

### Combinar condiciones

- `&` representa AND: ambas condiciones deben cumplirse.
- `|` representa OR: al menos una condición debe cumplirse.
- Encierra cada condición entre paréntesis.


```python
ventas_entre_100_y_200 = ventas[(ventas >= 100) & (ventas <= 200)]
ventas_extremas = ventas[(ventas < 100) | (ventas > 200)]

print("Ventas entre 100 y 200, inclusive:", ventas_entre_100_y_200)
print("Ventas inferiores a 100 o superiores a 200:", ventas_extremas)
```

En datos numéricos, `np.nan` representa un valor faltante. Algunas funciones, como `np.mean`, propagan el `NaN` y devuelven `NaN`; otras, como `np.nanmean`, lo ignoran al calcular el promedio.

Importante: un cero no equivale a un dato faltante. Puede representar una venta real de cero unidades.


```python
ventas_incompletas = np.array([120, 150, np.nan, 200, 180, np.nan])

print("Máscara de valores faltantes:", np.isnan(ventas_incompletas)) #identificar valores faltantes
print("Cantidad de valores faltantes:", np.sum(np.isnan(ventas_incompletas)))
print("Promedio normal:", np.mean(ventas_incompletas))
print("Promedio ignorando NaN:", np.nanmean(ventas_incompletas))
```

El parámetro `axis` permite especificar la dirección de una agregación en un array multidimensional.

## 9. Identificar valores potencialmente atípicos

El rango intercuartílico (IQR) es una técnica exploratoria para detectar valores que se alejan del resto de las observaciones.

Se calcula:
- `IQR = Q3 - Q1`
- Límite inferior: `Q1 - 1.5 * IQR`
- Límite superior: `Q3 + 1.5 * IQR`

Los valores fuera de estos límites se consideran potencialmente atípicos, pero no necesariamente son errores.


```python
ventas_dia = np.array([100, 110, 115, 120, 130, 125, 900])

q1 = np.percentile(ventas_dia, 25)
q3 = np.percentile(ventas_dia, 75)
iqr = q3 - q1

limite_inferior = q1 - 1.5 * iqr
limite_superior = q3 + 1.5 * iqr

atipicos = ventas_dia[
    (ventas_dia < limite_inferior) |
    (ventas_dia > limite_superior)
]

print("Q1:", q1)
print("Q3:", q3)
print("IQR:", iqr)
print("Límite inferior:", limite_inferior)
print("Límite superior:", limite_superior)
print("Valores potencialmente atípicos:", atipicos)
```
