LIMPIEZA DE TEXTOS

Los más importantes:

.str.strip() → elimina espacios al inicio y final

.str.lower() → minúsculas

.str.upper() → mayúsculas

.str.title() → Iniciales en mayúscula

.str.replace() → reemplazar texto

```python
df["ciudad"].value_counts() #contar las celdas iguales 
```

```python
df["ciudad"] = (
    df["ciudad"]
    .str.strip()
    .str.lower()
    .str.title()
)

df["ciudad"].value_counts()
```

```python
df["ciudad"] = (
    df["ciudad"]
    .str.replace("Medellin", "Medellín", regex=False)
    .str.replace("Bogota", "Bogotá", regex=False)
)
#La funcion sirve para reemplazar medellin a como Medellín

df["ciudad"].value_counts()
```

```python
df["categoria"].value_counts()
```

```python
df["categoria"] = (
    df["categoria"]
    .str.strip()
    .str.lower()
    .str.title()
)

df["categoria"].value_counts()
```

```python
df["categoria"] = (
    df["categoria"]
    .str.replace("Tecnología", "Tecnologia", regex=False)
    .str.replace("Papelería", "Papeleria", regex=False)
)

df["categoria"].value_counts()
```

PRECIO UNITARIO

```python
df["precio_unitario"].head(10)
```

```python
df["precio_unitario"] = (
    df["precio_unitario"]
    .astype("string") # .astype () - convierte columna a texto
    .str.replace("$", "", regex=False)
    .str.replace(".", "", regex=False) # regex=False = busca exactamente ese texto, sin interpretarlo como una expresión regular
)

df["precio_unitario"].head(10)
```

Convertir a número

pd.to_numeric() intenta convertir los valores a un tipo numérico.

Con errors="coerce" los valores que no puedan convertirse pasan a ser NaN

```python
df["precio_unitario"] = pd.to_numeric(
    df["precio_unitario"],
    errors="coerce"
)

df["precio_unitario"].dtype
```

```python
df.info()
```

```python
df["descuento"].head(10)
```

```python
df["descuento"] = (
    df["descuento"]
    .astype("string") #se convierte la columna a texto
    .str.replace("%", "", regex=False)
)

df["descuento"] = pd.to_numeric( #se pasa a valor numerico
    df["descuento"],
    errors="coerce"
)
```

Ahora tenemos los porcentajes como números.

Para trabajar con el descuento en cálculos, se puede convertir a proporción:

10 → 0.10

5 → 0.05

```python
df["descuento"] = df["descuento"] / 100

df["descuento"].head(10)
```

Valores inválidos

Un dato puede tener el tipo correcto y aun así estar mal.

Ejemplo:

edad_cliente = 150
cantidad = -3

```python
df[df["edad_cliente"] < 18][["id_venta", "cliente", "edad_cliente"]].head(20)
```

```python
df[
    (df["edad_cliente"] < 18) |
    (df["edad_cliente"] > 100)
][["id_venta", "cliente", "edad_cliente"]].head(20)
```

```python
condicion_edad = (
    (df["edad_cliente"] < 18) |
    (df["edad_cliente"] > 100)
)

df.loc[condicion_edad, "edad_cliente"] = pd.NA
```

```python
df[df["edad_cliente"].isnull()]
```

```python
df[df["cantidad"] <= 0][
    ["id_venta", "producto", "cantidad"]
]
```
