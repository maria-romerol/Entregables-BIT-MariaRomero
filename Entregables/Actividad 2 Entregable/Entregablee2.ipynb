## Estandarizar columnas de tipo de animales y países

```python
da[da["Animal type"].isnull()] #Solo se puede recuperar 8 datos que contienen informacion de Austria y vamos a comparar el peso que se pone es un animal pequeño
#puede ser red squirrel o hedgehog           
```

```python
da.dropna(subset=['Animal type', 'Country'], how='all', inplace=True) #con esto eliminamos las filas segun condiciones de columnas
```

```python
#Normalizar los nombres, sin espacios, con mayusculas
da["Animal type"] = (
    da["Animal type"]
    .str.strip()
    .str.lower()
    .str.title()
)
#Quitar los valores TM y ?
da["Animal type"] = da["Animal type"].str.replace('™', '', regex=False)
da["Animal type"] = da["Animal type"].str.replace('?', '', regex=False)
da["Animal type"] = da["Animal type"].str.strip()

#vamos a hacer un diccionario para las correcciones
correccion_A = {
    'Red Squirrell': 'Red Squirrel',
    'Red Squirel': 'Red Squirrel',
    'European Bisson': 'European Bison',
    'European Buster': 'European Bison',
    'Ledgehod': 'Hedgehog',
    'Wedgehod': 'Hedgehog'
}

da["Animal type"] = da["Animal type"].replace(correccion_A)
da["Animal type"].value_counts()
#Me quedan pendiente 8 valores que están NaN en Austria
```

```python
da[(da["Country"].isin(["Australia", "Austria"])) & (da["Animal type"].str.contains("Red Squirrel", case=False, na=False))] 
#Al comparar nos damos cuenta con la latitud y longitud que no hay datos para Australia, son para Austria
```

```python
da["Country"] = (
    da["Country"]
    .str.strip()
    .str.lower()
    .str.title()
)

correccion_P = {
    'Pl': 'Poland',
    'Hu': 'Hungary',
    'Hungry': 'Hungary',
    'Cz': 'Czech Republic',
    'Czech': 'Czech Republic',
    'De': 'Germany',
    'Cc': 'Czech Republic',
    'Australia': 'Austria'
}
da["Country"] = da["Country"].replace(correccion_P)
da["Country"].value_counts()
```

```python
L4 = (da["Animal type"] == "Red Squirrel") & (da["Country"] == "Germany")
nulos = L4 & da["Latitude"].isnull()

lat_media = da.loc[L4, "Latitude"].mean()
lon_media = da.loc[L4, "Country"].eq("Germany") 
lon_media = da.loc[L4, "Longitude"].mean()

da.loc[nulos, "Latitude"] = lat_media + np.random.uniform(-0.05, 0.05, size=nulos.sum())
da.loc[nulos, "Longitude"] = lon_media + np.random.uniform(-0.05, 0.05, size=nulos.sum())
```

```python
porcentaje_adultos = np.mean(da['Life_Stage'] == 'Adulto') * 100
porcentaje_jovenes = np.mean(da['Life_Stage'] == 'Joven') * 100

print(f"Porcentaje de adultos: {porcentaje_adultos:.2f}%")
print(f"Porcentaje de jóvenes: {porcentaje_jovenes:.2f}%")
```

```python
da['Life_Stage'] = np.where(
    da['Animal type'] == 'Red Squirrel', 
    np.where(da['Weight kg'] > 0.250, 'Adulto', 'Joven'),
    np.where(
        da['Animal type'] == 'European Bison', 
        np.where(da['Weight kg'] > 300, 'Adulto', 'Joven'),
        np.where(
            da['Animal type'] == 'Hedgehog', 
            np.where(da['Weight kg'] > 0.500, 'Adulto', 'Joven'),
            np.where(
                da['Animal type'] == 'Lynx', 
                np.where(da['Weight kg'] > 15, 'Adulto', 'Joven'),
                'No especificado'  # Valor por defecto si no coincide con ninguno
            )
        )
    )
)
```
