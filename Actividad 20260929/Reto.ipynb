```python
nombre = "Portatil"
precio = 1200000
cantidad = 15
disponible = True

#Impuesto
#costo_total= precio * 0.19
#correcion profe
subtotal = precio * cantidad
impuesto = subtotal * 0.19
costo_total= subtotal + impuesto

#Stock
stock = disponible and cantidad > 0

print(f"Nombre del producto: {nombre}")
print(f"Precio sin IVA: {precio}")
print(f"Cantidad en bodega: {cantidad}")
print(f"Disponibilidad del producto: {disponible}")
print(f"Costo total (+IVA): {costo_total}")
print(f"¿Hay stock disponible y es mayor a 0? {stock}")
print(type(nombre))
print(type(precio))
print(type(cantidad))
print(type(disponible))
print(type(costo_total))
print(type(stock))

#correccion profe
print(f"Nombre: {nombre}, tipo de dato {type(nombre)}")
print(f"Precio: {precio}, tipo de dato {type(precio)}")
print(f"Precio: {cantidad}, tipo de dato {type(cantidad)}")
print(f"Disponibilidad: {disponible}, tipo de dato {type(disponible)}")
print(f"Costo total: {costo_total}, tipo de dato {type(costo_total)}")
print(f"Stock: {stock}, tipo de dato {type(stock)}")
```
