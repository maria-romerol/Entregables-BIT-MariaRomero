## Limpieza de Datos

```python
#Se eliminan valores duplicados, se va a eliminar la columna de "Animal code" ya que no tienen valores útiles para este análisis
da = da.drop_duplicates()
da=da.drop(columns=['Animal code'])
da=da.drop(columns=['Animal name'])


print(da.shape)
da.info()
```

```python
da.isna().sum()
```

## Estandarizar columnas de tipo de animales y países

```python
da[da["Animal type"].isnull()] #Solo se puede recuperar 8 datos que contienen informacion de Austria y vamos a comparar el peso que se pone es un animal pequeño
#puede ser red squirrel o hedgehog           
```

```python
da.dropna(subset=['Animal type', 'Country'], how='all', inplace=True) #con esto eliminamos las filas segun condiciones de columnas
```

```python
#Normalizar los nombres, sin espacios, con mayusculas
da["Animal type"] = (
    da["Animal type"]
    .str.strip()
    .str.lower()
    .str.title()
)
#Quitar los valores TM y ?
da["Animal type"] = da["Animal type"].str.replace('™', '', regex=False)
da["Animal type"] = da["Animal type"].str.replace('?', '', regex=False)
da["Animal type"] = da["Animal type"].str.strip()

#vamos a hacer un diccionario para las correcciones
correccion_A = {
    'Red Squirrell': 'Red Squirrel',
    'Red Squirel': 'Red Squirrel',
    'European Bisson': 'European Bison',
    'European Buster': 'European Bison',
    'Ledgehod': 'Hedgehog',
    'Wedgehod': 'Hedgehog'
}

da["Animal type"] = da["Animal type"].replace(correccion_A)
da["Animal type"].value_counts()
#Me quedan pendiente 8 valores que están NaN en Austria
```

```python
da[(da["Country"].isin(["Australia", "Austria"])) & (da["Animal type"].str.contains("Red Squirrel", case=False, na=False))] 
#Al comparar nos damos cuenta con la latitud y longitud que no hay datos para Australia, son para Austria
```

```python
da["Country"] = (
    da["Country"]
    .str.strip()
    .str.lower()
    .str.title()
)

correccion_P = {
    'Pl': 'Poland',
    'Hu': 'Hungary',
    'Hungry': 'Hungary',
    'Cz': 'Czech Republic',
    'Czech': 'Czech Republic',
    'De': 'Germany',
    'Cc': 'Czech Republic',
    'Australia': 'Austria'
}
da["Country"] = da["Country"].replace(correccion_P)
da["Country"].value_counts()
```

```python
da[(da["Animal type"]=="Red Squirrel")].describe() #Con esto nos damos cuenta que los 8 valores faltantes en tipo de animal corresponden a la categoria "Red squirrel"
```

```python
#Rellenamos los valores nulos con Red squirrel
da['Animal type']=da['Animal type'].fillna('Red Squirrel')
da["Animal type"].value_counts()
```

# Analisis para especie Red Squirrel

```python
da["Animal type"].value_counts()
```

```python
da[da["Animal type"]=="Red Squirrel"]["Weight kg"].describe()
```

```python
#Se saca la media para hembras y machos, estos son los valores que vamos a rellenar en los valores faltantes y los valores negativos
#Hembra
Filtroh = (da['Animal type'] == "Red Squirrel") & (da['Gender'] == "Female") & (da['Weight kg'] > 0)
mediah = np.mean(da[Filtroh]['Weight kg'])
print(mediah)
#Macho
Filtrom = (da['Animal type'] == "Red Squirrel") & (da['Gender'] == "Male") & (da['Weight kg'] > 0)
mediam = np.mean(da[Filtrom]['Weight kg'])
print(mediam)
```

```python
#Vamos a pasar los valores negativos a nulos
da.loc[da['Weight kg'] < 0, 'Weight kg'] = np.nan
```

```python
#Ahora vamos a llenar los valores nulos por la media
da.loc[Filtroh,'Weight kg']=da.loc[Filtroh,'Weight kg'].fillna(mediah)

da.loc[Filtrom,'Weight kg']=da.loc[Filtrom,'Weight kg'].fillna(mediam)

da[da["Animal type"]=="Red Squirrel"]["Weight kg"].describe()
```

```python
#Ahora para el Body Length cm
da[da["Body Length cm"].isna() | (da["Body Length cm"]<0)]
#Con esto vemos que hay valores negativos por lo que podríamos quitar estos valores y ver el mínimo real
```

```python
da.loc[da['Body Length cm'] < 0, 'Body Length cm'] = np.nan
da[da["Animal type"]=="Red Squirrel"]["Body Length cm"].describe()
#podemos llenar los valores faltantes con la media
```

```python
#hembra
Filtrohl = (da['Animal type'] == "Red Squirrel") & (da['Gender'] == "Female")
mediahl = np.mean(da[Filtrohl]['Body Length cm'])
print(mediahl)
#Macho
Filtroml = (da['Animal type'] == "Red Squirrel") & (da['Gender'] == "Male")
mediaml = np.mean(da[Filtroml]['Body Length cm'])
print(mediaml)

da.loc[Filtrohl,'Body Length cm']=da.loc[Filtrohl,'Body Length cm'].fillna(mediahl)

da.loc[Filtroml,'Body Length cm']=da.loc[Filtroml,'Body Length cm'].fillna(mediaml)
```

```python
da[da["Animal type"]=="Red Squirrel"]["Body Length cm"].describe()
```
