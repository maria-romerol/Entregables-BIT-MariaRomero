Filtrar información

```python
# definir bajo una condicion o palabra de las columnas ciudades nombre cualquier cosa

clientes[clientes["ciudad"] == "Cali"]
```

```python
#Se usa para cuando no se sabe con exactitud como se escribio la info, si usan tilde o no
clientes[clientes['ciudad'].str.startswith("Pe")] # Startswith - Empezar con tal letra
```

```python
#Se usa para cuando no se sabe con exactitud como se escribio la info, si usan tilde o no
clientes[clientes['nombre_cliente'].str.contains("Ju")] # Contains - contiene
```

```python
#Se usa para cuando no se sabe con exactitud como se escribio la info, si usan tilde o no
clientes[clientes['nombre_cliente'].str.endswith("ez")] # Endswith - Termine en tal letra
```

CON CONDICIONALES +2

```python
# Dos condiciones
#Traer la base a analizar
clientes[
    (clientes["segmento"]== "Regular") & # & = and-y
    (clientes["ciudad"]=="Manizales") | # | = or-o
    (clientes["nombre_cliente"].str.startswith("Lu"))
]
```

```python
clientes_filtrados = [nombre_cliente for nombre_cliente in clientes["nombre_cliente"]
                    if "Luis" in nombre_cliente]

clientes_filtrados
```
