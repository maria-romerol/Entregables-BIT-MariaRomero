Operadores Relacionales

```python
igualdad = "hola" == "adios"
desigualdad = 5 != 6
menor_que = 5 < 6
menor_o_igual_que = 5 <= 6
mayor_que = 5 > 6
mayor_o_igual_que = 5 >= 6
# Aqui aplica la logica, si cumple o no
```

Operaciones de Asignacion

```python
asignacion = 10 #se le da a todo
print(f"inicia: {asignacion}") #la f sirve para unir str con variables
#suma_asignacion
asignacion += 5
print(f"suma y asignacion: {asignacion}")
#asignacion = asignacion + 5 (+=)
print(f"suma y asignacion: {asignacion}")
#resta_asignacion
asignacion -= 5 #(15 - 5 = 10)
print(f"resta y asignacion: {asignacion}")
#multiplicacion_asignacion
# asignacion = 10
asignacion *= 5
print(f"multiplicacion y asignacion: {asignacion}")
#division_asignacion
asignacion /= 5
print(f"division y asignacion: {asignacion}")
#exponente_asignacion
asignacion **= 5.0
print(f"exponente y asignacion: {asignacion}")
#modulo_asignacion
asignacion %= 5

asignacion = asignacion % 5


print(f"modulo y asignacion: {asignacion}")
```

Operadores Logicos

```python
conjuncion = True and False #and si se cumplen todas las condiciones
disyuncion = True or False # or si se cumple aunque sea 1 condicion
negacion = not True # not invertir el valor
```

```python
print("conjuncion:", conjuncion)
print("disyuncion:", disyuncion)
print("negacion:", negacion)
```

```python
## EJEMPLO

comparacion_1 = 10 > 5 # True
comparacion_2 = 10 > 11 # False

print(comparacion_1 or comparacion_2)
print(comparacion_1 and comparacion_2)
```
