## 10. Reto final: análisis de ventas

### Contexto de negocio

Una tienda registra las unidades vendidas durante diez días. El gerente quiere conocer el desempeño comercial, identificar días críticos y establecer una alerta para ventas bajas.

```python
ventas = np.array([120, 85, 150, 200, 95, 180, 75, 220, 160, 90])
```

### Tareas

Resuelve las siguientes preguntas usando NumPy:

1. Calcula el total de unidades vendidas.
2. Calcula el promedio y la mediana.
3. Identifica los días con ventas inferiores a 100.
4. Calcula el porcentaje de días con ventas superiores a 150.
5. Usa `np.where()` para asignar la etiqueta `"ALERTA"` a las ventas inferiores a 100 y `"NORMAL"` al resto.
6. Escribe dos conclusiones que puedan ser útiles para el gerente.




````python
import numpy as np
````

````python
#1. Calcula el total de unidades vendidas.
ventas = np.array([120, 85, 150, 200, 95, 180, 75, 220, 160, 90])
print("Total de unidades vendidas:", np.sum(ventas))

#2. Calcula el promedio y la mediana.
print("Promedio de ventas:", np.mean(ventas)) 
print("Mediana:", np.median(ventas))

#3. Identifica los días con ventas inferiores a 100.
venta_menor = ventas<100
print("Dias con ventas inferiores a 100:", np.sum(venta_menor))
dia = 1 
for i in venta_menor:
    if i == True:
        print(f"El día {dia} tuvo ventas inferiores a 100.")
    dia += 1

#4. Calcula el porcentaje de días con ventas superiores a 150.
venta_superior = ventas>150
porcentaje= np.mean([venta_superior])
print("Porcentaje de días con ventas superiores a 150:", porcentaje * 100,"%")

#5. Usa `np.where()` para asignar la etiqueta `"ALERTA"` a las ventas inferiores a 100 y `"NORMAL"` al resto.
etiqueta= np.where(ventas<100, "Alerta", "Normal")
etiqueta_matriz = etiqueta.reshape(2, 5)
print(etiqueta_matriz)

#6. Escribe dos conclusiones que puedan ser útiles para el gerente.
# 1. Con el promedio de 137.5 y la mediana de 135.0, es posible generar un objetivo de ventas mínimas por día de 100 unidades.
# 2. Se evidencian 4 días de los 10 días evaluados con alertas, por lo que es necesario conocer qué factores están causando estas bajas, ya que representa un 40%,
# así como evaluar la estabilidad del mercado ya que después de una venta < 100 und, al día siguiente se evidencia un aumento significativo en ventas.
````
