```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.DataFrame({
    "Mes": ["Ene", "Feb", "Mar", "Abr", "May", "Jun"],
    "Ventas": [120, 150, 135, 180, 210, 195],
    "Categoria": ["A", "B", "A", "C", "B", "C"],
    "Unidades": [20, 35, 25, 40, 30, 45]
})

display(df)
```

## 2. Primer desafío: Matplotlib

Crea un gráfico de líneas que represente las ventas por mes.

**Requisitos:**

* Agrega un título y nombres a los ejes.
* Personaliza el color de la línea y los marcadores.
* Cambia el estilo de la línea.
* Agrega una cuadrícula y ajusta el tamaño de la figura.

```python
plt.figure(figsize=(7, 4)) ## Tamaño de las figura
plt.plot(df['Mes'], df['Ventas'], marker='*', color='pink', linewidth=1, linestyle='-.') ## eje x, eje y
plt.title("Ventas por Mes")
plt.xlabel("Mes")
plt.ylabel("Ventas")
plt.xticks(df['Mes'], rotation=0) # Ajuste en los nombres del eje x, rotarlos para que se puedan leer
plt.grid()
plt.show()
```

## 3. Segundo desafío: Seaborn

Crea un gráfico de barras que muestre las unidades por categoría.

**Requisitos:**

* Selecciona una paleta de colores diferente a la predeterminada.
* Cambia el estilo visual del gráfico.
* Personaliza el título y las etiquetas de los ejes.
* Elige colores que faciliten la lectura.

```python
sns.set_style('white')
plt.figure(figsize=(10, 5))
sns.barplot(data=df, x="Categoria", y="Unidades", palette = "RdPu", 
            hue='Categoria', legend= True)
plt.title("Unidades por Categoria")
plt.xlabel('Categoria')
plt.ylabel('Unidades')
plt.show()
```
