```python
#añadir nuevo producto 
inventario.append({"producto": "Abrigo", "precio": 50000, "stock": "2"})
print(inventario)

#convertir valores de stock en int
inventario[0]["stock"]= int(inventario[0]["stock"]) 
inventario[1]["stock"]= int(inventario[1]["stock"]) 
inventario[2]["stock"]= int(inventario[2]["stock"]) 

print(type(inventario[0]["stock"]))
print(type(inventario[1]["stock"]))
print(type(inventario[2]["stock"]))

#aumentar 10000 a cada precio
inventario[0]["precio"] += 10000
inventario[1]["precio"] += 10000
inventario[2]["precio"] += 10000
print(inventario)

#imprimir cada producto
for item in inventario:
    stock = item["stock"]
    producto = item["producto"]
    precio = item["precio"]
    print(f"Hay {stock} unidades del producto {producto}. Su precio por unidad es de {precio}")
```
